# Feature Encoding


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import preprocessing
```

```python
df = pd.read_csv('/content/Churn_Modelling.csv')
```

```python
df.head(5)
```

```python
df.info()
```

```python
df.drop(columns = ["CustomerId","RowNumber","Surname"], axis = 1, inplace=True)
```

```python
df.head(5)
```

```python
df.Gender.value_counts()
```

Handling Missing values in Gender

```python
df.Gender.mode()
```

```python
df['Gender'] = df['Gender'].fillna('Male')
```

```python
df.Gender.value_counts()
```

```python
df.info()
```

Label Encoding

```python
le = preprocessing.LabelEncoder()
df['Gender_label'] =le.fit_transform(df.Gender.values)
```

```python
df.head(5)
```

```python
df.Gender_label.value_counts()
```

One Hot Encoding

```python
one_hot = pd.get_dummies(df['Geography'])
```

```python
one_hot
```

```python
df.head(5)
```

```python
df_dummies = pd.get_dummies(df)

df_dummies.head(5)
```

###Dummy Encoding

```python
df_dummies_de = pd.get_dummies(df, drop_first=True)

df_dummies_de.head(5)
```
